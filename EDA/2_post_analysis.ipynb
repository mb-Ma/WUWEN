```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
# load prediction and label
data_gru = np.load("../outputs/2025-09-03/16-18-38/result.npz")
data_nbeats = np.load("../outputs/2025-09-03/22-01-44/result.npz")
```

```python
x_axis = np.arange(12)
```

```python
fig, axes = plt.subplots(3, 4, figsize=(16, 12))

for i in range(3):
    for j in range(4):
        num = np.random.randint(0, data_gru['real_y'].shape[0], size=1)[0]
        sns.lineplot(x=x_axis, y=data_gru['real_y'][num, :, 0], ax=axes[i, j], label='true')
        sns.lineplot(x=x_axis, y=data_gru['pred_y'][num,:, 0], ax=axes[i, j], label='gru')
        sns.lineplot(x=x_axis, y=data_nbeats['pred_y'][num,:, 0], ax=axes[i, j], label='nbeats')
        axes[i, j].set_title(str(num))

# 调整布局
plt.tight_layout()
plt.subplots_adjust(top=0.93)
plt.show()
```
